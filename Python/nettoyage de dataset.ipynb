```python
print(dfM['delai_saisie_min'].describe())
```

```python
#cette fonction permter de faire une segmentation de delai de saisir de date par le chauffeur
def categorie_remontee(x):
    if pd.isna(x):
        return 'Non calculable'
    elif x < 0:
        return 'Anomalie'
    elif x < 5:
        return '< 5 min'
    elif x < 10:
        return '5 à 10 min'
    elif x < 60:
        return '10 à 60 min'
    else:
        return '> 60 min'

dfM['categorie_remontee'] = dfM['delai_saisie_min'].apply(categorie_remontee)
```
